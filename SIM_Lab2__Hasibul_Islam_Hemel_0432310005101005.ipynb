```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
ID = "0432310005101005"
d1 = int(ID[-1])
d2 = int(ID[-2])

A = np.array([
    [d1+2 , d2+1],
    [2*d1, d2+2]
])
print("matrix A: \n", A)
size_A = A.shape
det_A = np.linalg.det(A)
rank_A = np.linalg.matrix_rank(A)
eigenvalues_A, eigenvactors_A = np.linalg.eig(A)

print("\nsize of A:\n", size_A)
print("\ndet of A:\n", det_A)
print("\nrank of A:\n", rank_A)
print("\neigenvalues of A:\n", eigenvalues_A)
print("\neigenvactors of A:\n", eigenvactors_A)

if det_A != 0:
    inv_A = np.linalg.inv(A)
    print("Inverse of A is :\n", inv_A)
else:
    print("Inverse is not possible")
```

```python
# Value Changing exactly one element (+1)
B = A.copy()
B[1, 0] += 1
print("\nMatrix B after adding +1 into A[1, 0]):\n", B)
size_B = B.shape
det_B = np.linalg.det(B)
rank_B = np.linalg.matrix_rank(B)
eigenvalues_B, eigenvactors_B = np.linalg.eig(B)

print("\nsize_B:\n", size_B)
print("\ndet of B:\n", det_B)
print("\nrank of B:\n", rank_B)
print("\neigenvalues of B:\n", eigenvalues_B)
print("\neigenvactors of B:\n", eigenvactors_B)

if det_B != 0:
    inv_B = np.linalg.inv(B)
    print("Inverse of B is :\n", inv_B)
else:
    print("Inverse is not possible")
```



1.How did the determinant change and why?
   The determinant dropped from 8 to 3 after one entry of the matrix was altered. This reduction occurs because changing a single element affects the overall relationship between the rows of the matrix.

2.Did the rank change? Explain.
   No, the rank stayed the same at 2 for both matrices. This is because the rows remain linearly independent even after the modification.

3.How did the eigenvalues respond to the value change?
   The eigenvalues shifted from ([1, 8]) to approximately ([0.35, 8.65]). This indicates that eigenvalues can be quite sensitive to small changes in matrix entries.

4.Is B easier or harder to invert than A? Why?
   Matrix B is harder to invert since it has a smaller determinant. A lower determinant generally means the inverse is less stable and more sensitive to numerical errors.
